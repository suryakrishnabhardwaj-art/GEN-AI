```python
from google.colab import userdata
from openai import OpenAI

api_key = userdata.get("OPENAI_API_KEY")
if not api_key:
    raise ValueError("API key not found. Check Colab Secrets.")

client = OpenAI(api_key=api_key)
print("OpenAI client created and authenticated successfully!")
```

```python
prompt = """Explain Artificial Intelligence in simple words. Give five real-world examples suitable for a first-year engineering student."""

try:
    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[
            {"role": "user", "content": prompt}
        ]
    )

    print("========== LLM RESPONSE ==========")
    print(response.choices[0].message.content)
    print("==================================")
except Exception as e:
    print(f"An error occurred: {e}")
    print("Please verify that your OPENAI_API_KEY in Colab Secrets is correct.")
```

```python
from google import genai
from google.colab import userdata
import getpass

gemini_key = None

# Try retrieving GEMINI_API_KEY safely
try:
    gemini_key = userdata.get('GEMINI_API_KEY')
except Exception:
    pass

# Try retrieving OPENAI_API_KEY safely if GEMINI_API_KEY was not found
if not gemini_key:
    try:
        gemini_key = userdata.get('OPENAI_API_KEY')
    except Exception:
        pass

# Fallback to manual input if neither exists in Colab Secrets
if not gemini_key:
    print("Gemini key not found under GEMINI_API_KEY or OPENAI_API_KEY in Secrets.")
    gemini_key = getpass.getpass("Please paste your Gemini API Key here: ")

try:
    if gemini_key:
        client_gemini = genai.Client(api_key=gemini_key)
        print("Gemini client successfully created!")
    else:
        raise ValueError("API Key input was empty.")
except Exception as e:
    print(f"Error setting up Gemini client: {e}")
```

```python
prompt_gemini = """Explain Artificial Intelligence in simple words. Give five real-world examples suitable for a first-year engineering student."""

try:
    response = client_gemini.models.generate_content(
        model='gemini-3.5-flash',
        contents=prompt_gemini,
    )
    print("========== GEMINI RESPONSE ==========")
    print(response.text)
    print("=====================================")
except Exception as e:
    print(f"An error occurred while calling Gemini: {e}")
```
